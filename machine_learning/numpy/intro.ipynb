# Generator in Python

```python
from collections.abc import Generator 
def count_up_to_n(n) -> Generator[int]:
    count: int = 1 
    while count <= n:
        yield count
        count += 1
```

```python
for i in count_up_to_n(5):
    print(i)

count_up_to_n(5)
```

```python
def advanced_echo() -> Generator[str, int, bool]:
    value = yield "Ready"
    if value == 42:
        return True
    return False
```

```python
advanced_echo()
```

```python
g = advanced_echo()
print(next(g))  # start the generator, yields "Ready"
try:
    g.send(42)
except StopIteration as e:
    print(e.value)  # prints the returned value (True/False)
```

# Universal functions
- NumPy provides familiar mathematical functions such as sin, cos, and exp. In NumPy, these are called “universal functions” (ufunc). Within NumPy, these functions operate elementwise on an array, producing an array as output.



```python
# np.fromfunction() -> is used to generate elements based on the specific logic given by a callback function (cb, shape, dtype)
```

```python
# 1. Define a function with two parameters: 'row' and 'col'
def add_coordinates(row, col):
    print("Row parameter received:\n", row)
    print("Col parameter received:\n", col)
    return row + col

# 2. Call fromfunction with a shape of (2, 3)
result = np.fromfunction(add_coordinates, (2, 3), dtype=int)

print("\nFinal Result Array:\n", result)
```

```python
a = np.fromfunction(lambda x, y: 10 * x + y, (5,4), dtype="int64" )
```

```python
a
```

```python
for idx, row in enumerate(a):
    print(idx, row)
```

```python
# flat -> is the iterator over all elements
for ele in a.flat:
    print(ele)
```
