```python
# Check if a number is prime.
import math

num1 = int(input("Enter a number: "))
flag = False

if (num1 <= 1): print("not prime number")
else:
    for i in range(2, int(math.sqrt(num1)) + 1):
        if (num1 % i == 0):
            flag = True
            break
    
    if not flag:
        print("Prime number")
    else:
        print("not prime")
```

```python
# Reverse a number.

num1: int = int(input("Enter a number: "))
num = 0

while (num1 > 0) :
    remainder = num1 % 10
    num = num * 10 + remainder
    num1 = num1 // 10 

print(num)
```

```python
# Find factorial using loop.
num1: int = int(input("Enter a number: "))
fact = 1

while ( num1 > 0 ):
    fact = fact * num1
    num1 = num1 - 1

print(fact)
```

```python
# Find factorial using recursion.
num1: int = int(input("Enter a number: "))

def fact(num):
    if num <= 1: return 1
    return num * fact(num - 1)

print("fact of " , num1, " ", fact(num1))
```

```python
# Check palindrome string.


str1: str = str(input("Enter a string: "))
str_rev: str = ""

str_rev = str1[-1::-1]

if str1.lower() == str_rev.lower():
    print("palindrone")
else:
    print("not palindrone")
```

2. Strings (Important)

```python
# Find duplicate characters in a string.

str1: str = str(input("Enter a string: "))
duplicacy = {}

for i in str1:
    duplicacy[i] = duplicacy.get(i, 0) + 1

for key, count in duplicacy.items():
    if (count > 1):
        print(key, end=" ")
```

```python
# Find first non-repeating character.

str1: str = str(input("Enter a string: "))

for i in range(0, len(str1)):
    flag: bool = True
    for j in range(0, len(str1)):
        if (i == j): 
            continue;
        if (str1[i] == str1[j]):
            flag = False
            break
   
    if flag: 
        print(str1[i])
        break
```

```python
# Compress string:

str1: str = str(input("Enter a string: "))
object_cnt = {}
compress_str = ""
for i in str1:
    object_cnt[i] = object_cnt.get(i, 0) + 1

for ch, count in object_cnt.items():
    compress_str += ch + str(count)    

print(compress_str)
```

```python
# anagram string -> those words that are formed by the same letters by rearrangement. And using all the characters excatly once
# -> this imples that the length of both the words would be same.
# -> and the letters should be same.

str1: str = str(input("Enter String1: ")).lower()
str2: str = str(input("Enter String2: ")).lower()

flag: bool = True

if (len(str1) != len(str2)):
    print("Strings are not anagram")
else:
    str2_ = list(str2)
    for ch in str1:
        if ch in str2_:
            str2_.remove(ch)
        else: 
            flag = False
            break

    if flag == True:
        print("Strings are anagrams")
    else:
        print("Strings are not anagrams")
```
