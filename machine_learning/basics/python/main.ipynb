```python
# Find the largest among 3 numbers.

num1: int = int(input("Enter a number1: "))
num2: int = int(input("Enter a number2: "))
num3: int = int(input("Enter a number3: "))


if (num1 == num2 and num2 == num3):
    print("all numbers are same")
elif (num1 > num2):
    if (num1 > num3):
        print("num1 is greater")
    else:
        print("num3 is greater")
else:
    if (num2 > num3):
        print("num2 is greater")
    else:
        print("num3 is greater")
```

```python
# Check if a number is prime.
import math

num1 = int(input("Enter a number: "))
flag = False

if (num1 <= 1): print("not prime number")
else:
    for i in range(2, int(math.sqrt(num1)) + 1):
        if (num1 % i == 0):
            flag = True
            break
    
    if not flag:
        print("Prime number")
    else:
        print("not prime")
```

```python
# Reverse a number.

num1: int = int(input("Enter a number: "))
num = 0

while (num1 > 0) :
    remainder = num1 % 10
    num = num * 10 + remainder
    num1 = num1 // 10 

print(num)
```

```python
# Find factorial using loop.
num1: int = int(input("Enter a number: "))
fact = 1

while ( num1 > 0 ):
    fact = fact * num1
    num1 = num1 - 1

print(fact)
```

```python
# Find factorial using recursion.
num1: int = int(input("Enter a number: "))

def fact(num):
    if num <= 1: return 1
    return num * fact(num - 1)

print("fact of " , num1, " ", fact(num1))
```

```python
# Count vowels in a string.

str1: str = str(input("Enter a string: "))
cnt: int = 0

for i in range(0, len(str1)):
    if (str1[i].upper() in ["A", "E", "I", "O", "U"]):
        cnt +=1

# Nishidh
print("vowel count: ", cnt)
```

```python
# Check palindrome string.


str1: str = str(input("Enter a string: "))
str_rev: str = ""

str_rev = str1[-1::-1]

if str1.lower() == str_rev.lower():
    print("palindrone")
else:
    print("not palindrone")
```
