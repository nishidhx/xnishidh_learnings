```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing  import PolynomialFeatures
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
```

```python
X = np.random.rand(50, 1)
equation = 1.3 * (X ** 8) - \
    1.4 * (X ** 7) - \
    0.8 * (X ** 6) + \
    0.6 * (X ** 5) - \
    0.3 * (X ** 4) + 0.5 * (X ** 3) + 1.2 * (X ** 2) - 0.6 * X + 0.1 * np.random.rand(50, 1) 
```

```python
fig = plt.figure()
plt.scatter(X, equation)
plt.show()
```

```python
ig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.ravel()  # Flatten the 2D array of axes

for i in range(1, 9):
    poly = PolynomialFeatures(degree=i)
    X_poly = poly.fit_transform(X)

    scaler = StandardScaler()
    X_poly_std = scaler.fit_transform(X_poly)

    model = LinearRegression()
    model.fit(X_poly_std, equation)

    prediction = model.predict(X_poly_std)

    ax = axes[i - 1]
    ax.scatter(X, equation, label="Dataset", s=15)
    ax.scatter(X, prediction, label="Prediction", s=15)
    ax.set_title(f"Degree {i}")
    ax.grid(True)

# Show legend only once
axes[0].legend()

plt.tight_layout()
plt.show()
```
